```python
boroughs = ["Sutton", "Croydon", "Merton", "Kingston upon Thames", "Richmond upon Thames", "Wandsworth"]
```

```python
print(boroughs)
```

```python
print(boroughs[0]) #"Sutton", the first item
print(boroughs[1]) #"Croydon", the second item
print(boroughs[2]) #"Merton", the third item
print(boroughs[3]) #"Kingston upon Thames", the fourth item    
print(boroughs[4]) #"Richmond upon Thames", the fifth item
print(boroughs[5]) #"Wandsworth", the sixth item
print(boroughs[6]) #"Greenwich", the seventh item

```

```python
boroughs.append("Greenwich") #method: adds to the end
print(len(boroughs)) #function: how many items
```

```python
filenames = []

for borough in boroughs:
    clean_name = borough.lower().replace(" ", "_") #method: replaces spaces with underscores
    filename = f"{clean_name}.png" #f-string: format string, variable in
    filenames.append(filename) #method: adds to the end

print(filenames)
print(len(filenames)) #function: how many items
```

```python
filenames = ["sutton.png", "croydon.png", "merton.png", "kingston upon thames.png", "richmond upon thames.png", "wandsworth.png", "greenwich.png"]
for filename in filenames:
    text = filename.lower().replace(" ", "_") #method: replaces spaces with underscores
    print(f"File {filename} is an image of {text}")
 #f-string: format string, variable in string
```
