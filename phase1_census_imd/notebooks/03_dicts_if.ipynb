Dictionaries and IF Statements


```python
borough_status = {
    "Sutton": "outer",
    "Wandsworth": "inner",
    "Merton": "outer",
    "Kingston upon Thames": "outer",
    "Greenwich": "outer",
    "Richmond upon Thames": "outer",
    "City of London": "inner",
}
print(borough_status) #dictionary
print(borough_status["Sutton"]) #outer
print(borough_status["Wandsworth"]) #inner

```

```python
borough_status["Bromley"] = "outer"
print(borough_status)
 #add new key-value pair
```

```python
print(borough_status["Bromley"]) #outer
```

```python
print(len(borough_status)) #outer
```

```python
status = borough_status["Sutton"] #outer

if status == "inner":
    print(f"Inner London")
else: 
    print("Outer London")
```

```python
for borough, status in borough_status.items():
    if status == "inner":
        print(f"{borough} is an inner London borough")
    else:
        print(f"{borough} is an outer London borough")
```

```python
# 1. Create an empty list called inner_boroughs

# 2. Loop over borough_status, getting borough and status each time

    # 3. Inside the loop: check whether status is "inner"

        # 4. Inside the if: append the borough (not the status) to the list

# 5. After the loop (no indent): print the list

# 6. Print its length
```

```python
inner_boroughs = []
for borough, status in borough_status.items():
    if status == "inner":
        inner_boroughs.append(borough)
print(inner_boroughs)
print(len(inner_boroughs))
```
