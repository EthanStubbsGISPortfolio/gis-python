```python
borough_status = {
    "Sutton": "outer",
    "Wandsworth": "inner",
    "Merton": "outer",
    "Kingston upon Thames": "outer",
    "Greenwich": "outer",
    "Richmond upon Thames": "outer",
    "City of London": "inner",
}
print(borough_status) #dictionary
print(borough_status["Sutton"]) #outer
print(borough_status["Wandsworth"]) #inner
```

```python

def describe_borough(name, status):
    sentence = f"{name} is an {status} London borough"
    return sentence
```

```python

for name, status in borough_status.items():
    result = describe_borough(name, status)
    print(result)

```

```python
def get_inner_boroughs(status_dict):
    inner_boroughs = []
    for borough, status in status_dict.items():
        if status == "inner":
            inner_boroughs.append(borough)
    return inner_boroughs
```
